# 08 — Chatman–Manville Rule Comparison

Targeting rules (peak or mean statistic; one rate per zone or per zone and time band; the project's Seattle rule) run paired on the same demand draws and scored on welfare: consumer surplus + revenue - search cost. A small run; the full sweep is `outputs/tables/cm_sweep.parquet` (script 07 `--rebuild-cm-sweep`).

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from phillyparking.scenarios.cm_sweep import chatman_manville_sweep, paired_table, window_dependence
from phillyparking.viz.charts import chatman_manville_results_chart
results = chatman_manville_sweep(n_replications=5, rolls_forward_years=4)
results.head()
```

## Welfare terms by rule and start occupancy, last year (median over cells)

```python
last = results[results['year'] == results['year'].max()]
grouped = last.groupby(['start_occupancy', 'rule'])[
    ['welfare_change', 'cs_change', 'revenue', 'search_cost', 'mean_price']].median()
grouped
```

## Each comparison against its reference, per cell

`breakeven_minutes_median` is the search time at 95% occupancy on a 10-space block at which the two rules give equal welfare; below it the lower-priced rule gives more, above it the higher-priced one.

```python
paired = paired_table(results)
window_dependence(paired)
```

## Use the packaged chart

```python
fig = chatman_manville_results_chart(results)
plt.show()
```

## Reading the result

On cruising alone the peak rule wins by construction (a mean is never above a 95th percentile of the same block-hours), so the question is the breakeven: how costly must cruising be for the parkers a peak rule prices off blocks with space to be worth it. Read the paired table cell by cell; nothing is pooled.
