# 07 — Welfare & Incidence

The search cost of parkers under each scenario and a placeholder distributional incidence calculation.

```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from phillyparking.scenarios.runner import run_all_scenarios
from phillyparking.welfare.incidence import gini, simple_incidence_summary
results = run_all_scenarios()
```

## Annual search cost of parkers per scenario

The project's one cruising model (`welfare/search_cost.py`), scored at the config's level in minutes per arrival at a full block; the level is a stated assumption.

```python
pd.Series({name: float(res.search_cost['annual_search_cost_usd'].sum())
           for name, res in results.items()}, name='search_cost_usd')
```

## Synthetic decile-uniform incidence

```python
rng = np.random.default_rng(0)
n = 5000
incidence_df = pd.DataFrame({
    'household_id': range(n),
    'income_decile': rng.integers(1, 11, size=n),
    'parking_spend_usd': rng.gamma(shape=2.0, scale=120.0, size=n),
})
summary = simple_incidence_summary(incidence_df)
summary
```

## Gini on synthetic distribution

```python
g = gini(incidence_df['parking_spend_usd'].values)
print(f'Gini = {g:.3f}')
```
